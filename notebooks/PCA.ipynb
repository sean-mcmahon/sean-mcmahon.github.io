PCA, prinicple component analysis, is an unsupervised learning method which performs dimensionality reduction. It involves performing sinular value decomposition, where the eignvectors of the data are found. These eigen vectors define the principle vectors of the data, the directions through the feature space with most encompanse the feature space.
