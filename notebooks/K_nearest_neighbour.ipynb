K Nearest Neighbours, or KNNs, can be supervised or unsupervised, probably more commonly used in supervised machine learning tasks. For supervised learning the KNN is given the entire training set, and then at test time finds the k, number of datapoints closest to the test point, or points. For a classification task, a input data point is assigned to the class which is the most numerous of the k nearest points found. For regression the average of the k nearest points is used as the output. 

In an unsupervised task, a KNN simply returns the K nearest points. 

Under the hood KNNs often process the training data into a K-D Tree for efficient search. A KD tree is a binary search tree where the nodes are hyperplanes rather than integers, and the side the query point is on dictacts to which child of the current node you go to next is.
