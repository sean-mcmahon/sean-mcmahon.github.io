A random forest is an ensemble method combining multiple decision trees to make a more powerful model. This ensemble is built by training some number, set at run time, of decsion trees on random subsets of the data and on random subsets of the training features. Becuase each decision tree is trained on a different data, and sometimes different features, there is little chance of the model overfitting when more trees are added. 

The various trees can be used to determine the usefulness of certain features by evaluating the performance increase from using a feature on the relevant trees. 

You can also calculate the Out of Bag, OOB, error, to measure the random forests performance on a sudo validation set. To calculate OOB error, you evaluate each tree on the data it was not trained on and accumulate the error from all the trees.
